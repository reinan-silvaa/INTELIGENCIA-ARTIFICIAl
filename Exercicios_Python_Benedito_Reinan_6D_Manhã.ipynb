```python
# Exercicio 9
# Explicação:
# Nesse exercício, usei o input() para receber o número. Criei a variável "fatorial" valendo 1
# (porque se fosse 0, a multiplicação zeraria tudo). O for com range() percorre de 1 até o número digitado.
# O detalhe importante aqui é o "numero + 1" no range, já que o Python sempre ignora o último valor.
# A cada volta, multiplico o acumulado pelo "i" atual e, no final, exibo o resultado com print().

# Recebe o valor do usuário e já transforma em inteiro para podermos fazer as contas
numero = int(input("Digite um número: "))

# Inicio com 1 porque, na multiplicação, se começar com 0 vai zerar o resultado inteiro
fatorial = 1

# O loop vai do 1 até o número digitado. Esse +1 é essencial pro Python não parar um número antes!
for i in range(1, numero + 1):

    # A cada rodada do laço, atualizo a variável multiplicando o valor que ela já tem pelo 'i' da vez
    fatorial = fatorial * i

# Depois que o loop termina de fazer todas as multiplicações, mostro o resultado final na tela
print(fatorial)
```

```python
# Exercicio 10
# Explicação:
# Nesse exercício, usei o input() para saber quantos números da sequência o usuário quer ver.
# Iniciei as duas primeiras variáveis com 0 e 1, que são a base da sequência de Fibonacci.
# Usei o for para rodar o cálculo, somando sempre os dois últimos para achar o próximo.
# O detalhe mais importante é a atualização das variáveis no final do laço, que faz a "fila andar" para a próxima soma.

# Pede quantos termos o usuário quer ver e já converte pra inteiro
quantidade = int(input("Digite a quantidade de números da sequência: "))

# A sequência de Fibonacci sempre começa com esses dois valores
numero1 = 0
numero2 = 1

# Já imprimo os dois primeiros direto na tela para tirar eles do caminho
print(numero1)
print(numero2)

# O loop começa no 2 porque os dois primeiros (0 e 1) já foram impressos ali em cima
for i in range(2, quantidade):

    # A lógica principal: o próximo número é sempre a soma dos dois anteriores
    proximo = numero1 + numero2

    # Mostro na tela a operação matemática para ficar mais fácil de entender o que aconteceu
    print(numero1, "+", numero2, "=", proximo)

    # Aqui é onde a fila anda! O numero1 assume o valor do numero2, e o numero2 pega o novo resultado.
    # Assim, na próxima volta do loop, a soma vai usar os valores atualizados corretamente.
    numero1 = numero2
    numero2 = proximo
```

```python
# Exercicio 12
# Explicação:
# Nesse exercício, usei o input() para receber o texto e criei um dicionário vazio.
# A sacada aqui é usar o for para varrer cada caractere e usar o dicionário para ir guardando a contagem.
# O if/else faz o trabalho pesado: se a letra já existe lá, adiciono +1 na contagem.
# Se for a primeira vez que a letra aparece, eu crio ela no dicionário valendo 1.
# No final, uso outro for só para ler esse dicionário pronto e imprimir os resultados na tela.

# Recebe a palavra ou frase (lembrando que espaços e pontuações também contam como caractere)
texto = input("Digite uma palavra ou frase: ")

# Cria um dicionário vazio. Ele é perfeito pra isso porque vai guardar a letra (chave) e a quantidade (valor)
contador = {}

# Esse loop vai passar por cada caractere da string que o usuário digitou, um por um
for letra in texto:

    # O if checa se essa letra já foi registrada no nosso dicionário antes
    if letra in contador:
        # Se ela já está lá, eu apenas pego a quantidade atual dela e somo mais 1
        contador[letra] = contador[letra] + 1

    else:
        # Se cair no else, significa que é a primeira vez que o programa vê essa letra
        # Então, eu crio essa letra no dicionário e já digo que ela apareceu 1 vez
        contador[letra] = 1

# Agora que a contagem terminou, faço um novo loop, mas dessa vez percorrendo o dicionário que montei
for letra in contador:
    # Exibo o resultado mostrando a letra e, logo em seguida, o valor total que foi acumulado nela
    print(letra, ":", contador[letra])
```

```python
# Exercicio 13
# Explicação:
# Nesse exercício, fiz a ordenação "na mão" em vez de usar uma função pronta do Python (como o .sort()).
# Criei a lista e usei dois laços 'for' aninhados: o primeiro fixa um número, e o segundo compara ele
# com todos os próximos elementos da lista.
# Se o número atual for maior que o da frente, eles trocam de lugar. A sacada foi usar uma variável "aux"
# (auxiliar) para guardar o valor temporariamente e não perder a informação na hora de sobrescrever.

# Cria a lista com os números bagunçados que vamos organizar
lista = [8, 3, 15, 1, 10, 6]

# O primeiro for percorre os índices da lista (o len() serve para o loop saber o tamanho exato dela)
for i in range(len(lista)):

    # O segundo for sempre olha pra frente (por isso o 'i + 1'). Ele vai comparar o elemento 'i' com o resto da lista.
    for j in range(i + 1, len(lista)):

        # O if faz a pergunta: "O número da esquerda é maior que o da direita?". Se for, eles estão na ordem errada.
        if lista[i] > lista[j]:

            # Guardo o valor do elemento atual na variável 'aux' pra ele não ser apagado no próximo passo
            aux = lista[i]

            # Pego o número menor (que estava lá na frente) e jogo pra posição atual (mais à esquerda)
            lista[i] = lista[j]

            # Pego o número maior (que eu tinha salvo em 'aux') e jogo lá pra posição da direita
            lista[j] = aux

# Depois que os dois loops terminam de fazer todas as trocas, o print mostra a lista final arrumadinha
print(lista)
```

```python
# Exercício 14
# Explicação:
# A lógica aqui foi usar o conceito de "variável acumuladora". Primeiro, criei a lista com os valores.
# Depois, declarei a variável 'soma' valendo 0, o que é essencial para ter um ponto de partida limpo.
# Utilizei o laço for para iterar diretamente sobre os elementos da lista (o Python facilita muito isso,
# não precisa ficar controlando o índice). Dentro do loop, a variável 'soma' vai se atualizando,
# somando o valor que ela já tem com o número daquela rodada. Por fim, o print joga o total na tela.

# Lista declarada com os valores que serão processados
lista = [5, 10, 15, 20, 25]

# Inicializo com 0 (que é o elemento neutro da soma) para o cálculo não dar erro na primeira volta
soma = 0

# O for passa por cada item da lista automaticamente, sem precisar de range() ou len()
for numero in lista:

    # A cada repetição, a variável 'soma' pega o próprio valor que já acumulou e acrescenta o 'numero' da vez
    soma = soma + numero

# Fora do loop (depois que todas as somas terminaram), exibo o resultado consolidado
print("A soma dos elementos da lista é:", soma)
```

```python
# Exercício 15
# Explicação:
# Essa é a forma mais direta usando apenas os recursos embutidos da linguagem.
# A função sum() faz a soma de todos os valores e a função len() descobre o tamanho da lista.
# A média aritmética é exatamente isso: a soma total dividida pela quantidade de itens.

# Crio a lista de números
lista = [1, 2, 5, 6]

# O sum() soma tudo (14) e o len() conta os itens (4). O resultado de 14 / 4 vai para a variável 'media'
media = sum(lista) / len(lista)

print("A média é:", media)
```

```python
# Exercício 16
# Explicação:
# Aqui eu usei o fatiamento (slicing). Como uma string no Python funciona como uma lista de caracteres,
# o [::-1] basicamente diz para o programa pegar o texto inteiro e ler com um "passo" negativo,
# ou seja, de trás para frente. É o jeito mais limpo de inverter.

# Recebe o texto que o usuário digitar
texto = input("Digite o texto que deseja inverter: ")

# O macete do [::-1]: não passo índice de início nem de fim, só digo que o pulo é de -1
texto_invertido = texto[::-1]

# Mostra na tela como ficou a palavra ao contrário
print("Texto invertido:", texto_invertido)
```

```python
# Exercício 17
# Explicação:
# A primeira etapa é tratar o texto: o .lower() joga tudo para minúsculo e o .split()
# quebra a frase nos espaços, criando uma lista de palavras.
# Para a contagem, usei um dicionário. O macete aqui é usar o método .get(),
# que protege o código de dar "KeyError" se a palavra ainda não existir no dicionário.

# Recebe a frase e já padroniza tudo pra minúsculo e recorta nos espaços
texto = input("Digite uma frase ou texto: ")
palavras = texto.lower().split()

# Dicionário vazio onde vamos guardar as palavras (chave) e as quantidades (valor)
contagem = {}

# O loop vai olhar palavra por palavra da nossa lista
for palavra in palavras:
    # A mágica do .get(): tenta pegar a quantidade atual. Se a palavra for nova, devolve 0.
    # Aí é só somar + 1 em cima desse resultado!
    contagem[palavra] = contagem.get(palavra, 0) + 1

# Mostra o dicionário final com a frequência de cada palavra
print(contagem)
```

```python
#Exercicio 18
# Explicação:
# A ideia aqui foi usar uma função para encapsular a regra matemática.
# O operador de módulo (%) é perfeito para checar divisores: se der resto 0, a divisão é exata.
# O detalhe mais legal é o "return False" dentro do if, que funciona como um freio de mão:
# achou um divisor, ele aborta a função na hora, não precisa testar os próximos números.

def e_primo(numero):
    # Por regra matemática, 1, 0 e números negativos não são primos. Já mato essa charada de cara.
    if numero <= 1:
        return False

    # O range vai do 2 até o número anterior ao digitado (já que o range ignora o último valor)
    for i in range(2, numero):

        # Se a divisão do nosso número por qualquer "i" desse laço não tiver resto, achamos um divisor
        if numero % i == 0:
            # Como tem divisor além do 1 e dele mesmo, não é primo. Já retorno Falso e a função para aqui.
            return False

    # Se o loop terminar de rodar inteirinho e nunca cair naquele 'return False' ali de cima,
    # significa que o número passou no teste. Então, ele é primo!
    return True

# Pede um número inteiro para o usuário
entrada = int(input("Digite um número inteiro: "))

# Chama a função passando a "entrada" que o usuário digitou.
# Como a função retorna True ou False, posso jogar ela direto dentro do if!
if e_primo(entrada):
    print(f"O número {entrada} é primo!")
else:
    print(f"O número {entrada} não é primo.")
```

```python
# Exercicio 16
# Explicação:
# Professor, aqui eu criei a função jogar_adivinha. Usei a biblioteca random pra
# sortear o número de 1 a 100. Como a pessoa pode chutar várias vezes, coloquei um
# while True pra segurar o loop até ela acertar e o código ler o break.
# Também coloquei um try/except na hora de ler o input, porque se a pessoa digitar
# alguma letra sem querer, o programa não trava, ele só avisa do erro e pede de novo.

import random

def jogar_adivinha():
    print("--- Jogo da Adivinhação ---")
    print("Tente adivinhar o número entre 1 e 100")
    print("---------------------------")

    # sorteia o número que a pessoa tem que adivinhar
    numero_secreto = random.randint(1, 100)
    tentativas = 0

    # loop infinito, só vai sair quando acertar (usando o break)
    while True:
        try:
            # pega o chute do usuário e já converte pra número inteiro
            palpite = int(input("\nQual o seu palpite? "))
            tentativas += 1

            # verifica se o usuário acertou na mosca
            if palpite == numero_secreto:
                print(f"Aí sim! Você acertou o número {numero_secreto} em {tentativas} tentativas!")
                break # mata o loop

            # se errou, o programa dá uma dica
            elif palpite < numero_secreto:
                print("Quase... tenta um número MAIOR.")
            else:
                print("Passou... tenta um número MENOR.")

        # se o usuário digitar uma letra ao invés de número, o except segura o erro pro código não quebrar
        except ValueError:
            print("Opa, entrada inválida! Digita só números mesmo.")

# chama a função pra rodar o jogo
jogar_adivinha()
```

```python
# Exercicio 20
# Explicação:
# A ideia aqui é transpor a matriz, ou seja, transformar linha em coluna.
# Em vez de fazer dois laços "for" soltos, usei "list comprehension" pra resolver
# tudo numa linha só. O macete foi usar len(matriz[0]) para descobrir exatamente
# quantas colunas a matriz original tem, pra evitar erro de índice na hora de inverter.

# Matriz inicial no formato lista de listas (3 linhas e 2 colunas)
matriz = [
    [5, 6],
    [7, 8],
    [9, 10]
]

# Aqui é onde a inversão acontece. Lendo de trás pra frente fica mais claro:
# O "for j..." anda pela quantidade de colunas originais.
# O for de dentro pega o item correspondente de cada linha.
# Detalhe: o matriz[0] garante que o código saiba a quantidade exata de colunas!
transposta = [[linha[j] for linha in matriz] for j in range(len(matriz[0]))]

# Imprime o resultado: a matriz que era 3x2 agora virou 2x3
print(transposta)
```

```python
# Exercicio 21
# Explicação:
# A lógica principal da busca binária foi mantida (dividindo a lista pela metade).
# A diferença aqui é que usei o input() para deixar o código interativo,
# permitindo que quem estiver testando escolha o número.
# No final, coloquei um if/else simples só para dar uma resposta mais amigável
# caso a função retorne -1 (que significa que o número não está na lista).

def busca_binaria(lista, alvo):
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2
        chute = lista[meio]

        if chute == alvo:
            return meio

        elif chute > alvo:
            fim = meio - 1

        else:
            inicio = meio + 1

    return -1

# --- Testando a implementação ---

# A lista base (sempre ordenada!)
lista_ordenada = [10, 20, 30, 40, 50, 60, 70]

# Mostra a lista na tela para a pessoa saber quais opções ela tem
print("Lista disponível:", lista_ordenada)

# Pede o número e já converte para inteiro
numero_procurado = int(input("Digite o número que deseja buscar: "))

# Chama a função passando a lista e o número que o usuário digitou
posicao = busca_binaria(lista_ordenada, numero_procurado)

# Verifica o resultado para dar a resposta certa
if posicao != -1:
    print(f"Encontrado! O número {numero_procurado} está no índice: {posicao}")
else:
    print(f"O número {numero_procurado} não existe nessa lista.")
```

```python
# Exercício 22 - Cifra de César
# Explicação:
# Nesse exercício eu utilizei a Cifra de César para criptografar uma mensagem.
# Primeiro, o programa recebe a mensagem e o número de posições que cada letra
# deve ser deslocada no alfabeto. Depois, ele percorre cada letra da mensagem,
# transforma a letra em um código numérico e soma o deslocamento escolhido.
# Assim, é criada uma nova letra e a mensagem criptografada é formada.
# Os espaços e outros caracteres que não são letras permanecem iguais.

mensagem = input("Digite uma mensagem: ")  # Pede uma mensagem ao usuário

deslocamento = int(input("Digite o deslocamento: "))  # Pede quantas posições as letras serão deslocadas

criptografada = ""  # Cria uma variável para guardar a mensagem criptografada

for letra in mensagem:  # Percorre cada letra da mensagem
    if letra.isalpha():  # Verifica se o caractere é uma letra
        codigo = ord(letra)  # Converte a letra para um código numérico
        nova_letra = chr(codigo + deslocamento)  # Cria uma nova letra com o deslocamento
        criptografada += nova_letra  # Adiciona a nova letra à mensagem
    else:  # Caso o caractere não seja uma letra
        criptografada += letra  # Mantém o caractere sem alteração

print("Mensagem criptografada:", criptografada)  # Mostra a mensagem criptografada
```

```python
# Exercício 23 - Descriptografar Cifra de César

# Explicação:
# Nesse exercício eu faço o processo contrário da criptografia realizada no exercício anterior.
# Primeiro, o programa recebe a mensagem que foi criptografada e o número de posições
# que foi utilizado no deslocamento. Depois, ele percorre cada letra da mensagem,
# transforma a letra em um código numérico e diminui o deslocamento informado.
# Dessa forma, a letra volta para a sua posição original no alfabeto.
# Os espaços e outros caracteres que não são letras permanecem iguais.
# No final, o programa mostra a mensagem descriptografada na tela.

mensagem = input("Digite a mensagem criptografada: ")  # Pede a mensagem que será descriptografada

deslocamento = int(input("Digite o deslocamento: "))  # Pede quantas posições as letras serão deslocadas

descriptografada = ""  # Cria uma variável para guardar a mensagem descriptografada

for letra in mensagem:  # Percorre cada letra da mensagem
    if letra.isalpha():  # Verifica se o caractere é uma letra
        codigo = ord(letra)  # Converte a letra para um código numérico
        nova_letra = chr(codigo - deslocamento)  # Volta a letra de acordo com o deslocamento
        descriptografada += nova_letra  # Adiciona a nova letra à mensagem
    else:  # Caso o caractere não seja uma letra
        descriptografada += letra  # Mantém o caractere sem alteração

print("Mensagem descriptografada:", descriptografada)  # Mostra a mensagem original
```

```python
# Exercício 25 - Classificação de Números
# Explicação:
# A lógica aqui é como separar itens em caixas diferentes.
# Primeiro, crio três listas vazias para representar as categorias.
# Depois, uso o laço for para passar por cada número da lista original.
# O bloco if/elif/else funciona como um filtro: dependendo da condição do número,
# o método .append() guarda ele na lista correspondente.

numeros = [12, -7, 0, 5, -2, 0, -15, 8]

# Cria as listas vazias (nossas "caixas") para guardar os resultados
positivos = []
negativos = []
zeros = []

# O loop varre todos os elementos da lista principal
for x in numeros:

    # Se for maior que zero, joga na lista de positivos
    if x > 0:
        positivos.append(x)

    # Se for menor que zero, joga na lista de negativos
    elif x < 0:
        negativos.append(x)

    # Se não caiu no if e nem no elif, obrigatoriamente é zero. Não precisa nem testar!
    else:
        zeros.append(x)

# Imprime cada lista para mostrar que os números foram separados corretamente
print("Positivos:", positivos)
print("Negativos:", negativos)
print("Zeros:", zeros)
```

```python
# Exercício 26 - Calculadora Simples
# Explicação:
# usei um dicionário
# Que Ele liga direto o número digitado à continha que precisa ser feita.
# Se digitar '1', o código já puxa a função de somar automaticamente. Fica muito mais limpo.
# O try/except continua ali embaixo só pra segurar a onda se a pessoa digitar uma letra
# no meio do cálculo, evitando que o programa feche na cara do usuário.

# 1. As funções que fazem as contas. Deixei tudo em uma linha só pra economizar espaço.
def somar(x, y): return x + y
def subtrair(x, y): return x - y
def multiplicar(x, y): return x * y
def dividir(x, y):
    # Se tentar dividir por zero, o código levanta um alerta antes de quebrar
    if y == 0:
        raise ZeroDivisionError("Não dá pra dividir por zero!")
    return x / y

# 2. O truque do dicionário: conecta a opção do menu direto na função
operacoes = {
    '1': somar,
    '2': subtrair,
    '3': multiplicar,
    '4': dividir
}

# 3. O loop que mantém a calculadora rodando
while True:
    print("\n--- CALCULADORA ---")
    print("1. Somar | 2. Subtrair | 3. Multiplicar | 4. Dividir | 5. Sair")

    opcao = input("O que você quer fazer? (1-5): ")

    if opcao == '5':
        print("Fechando a calculadora. Valeu!")
        break

    # Checa se o que a pessoa digitou tá nas opções do dicionário (1, 2, 3 ou 4)
    if opcao in operacoes:
        try:
            # Pede os números e já converte pra float (pra aceitar número com vírgula/ponto)
            num1 = float(input("Primeiro número: "))
            num2 = float(input("Segundo número: "))

            # Puxa a operação certa do dicionário e já manda os números pra ela
            resultado = operacoes[opcao](num1, num2)

            print(f"Resultado: {resultado}")

        except ValueError:
            print("Opa, deu erro! Digita só números mesmo.")
        except ZeroDivisionError as erro:
            print(f"Erro: {erro}")
    else:
        print("Opção inválida! Escolhe aí de 1 a 5.")
```

```python
# Exercício 28 - Gráfico de Barras
# Explicação:
# Usei um "while True" principal para eu poder ir digitando a fruta e o valor dela
# um por um. Quando eu quiser parar, é só digitar 'sair' e o break encerra a
# coleta de dados, indo direto desenhar o gráfico.
# No final, coloquei uma checagem rápida (len > 0) só pra garantir que o gráfico
# não tente abrir vazio se eu não tiver digitado nada.

import matplotlib.pyplot as plt

frutas = []
valores = []

print("--- Gráfico de Frutas ---")
print("Dica: Digite 'sair' no nome da fruta quando quiser gerar o gráfico.\n")

# Loop principal para coletar quantas frutas eu quiser
while True:
    fruta = input("Nome da fruta: ")

    # Se digitar sair, quebra o loop e vai desenhar o gráfico
    if fruta.lower() == 'sair':
        break

    # Loop interno só pra garantir que vou digitar um número válido pro valor
    while True:
        try:
            valor = float(input(f"Qual o valor para '{fruta}'? "))
            break
        except ValueError:
            print("Entrada inválida! Digita só números mesmo.")

    # Adiciono a fruta e o valor nas minhas listas
    frutas.append(fruta)
    valores.append(valor)

# Só vai montar o gráfico se tiver pelo menos 1 fruta na lista
if len(frutas) > 0:

    # 1. Criei a lista de posições [0, 1, 2...] baseada na quantidade de frutas
    xs = [i for i, _ in enumerate(frutas)]

    # 2. Desenhei as barras usando as posições X e os valores
    plt.bar(xs, valores, color="skyblue", edgecolor="black")

    # 3. Coloquei os títulos e os rótulos dos eixos
    plt.title("Meu Gráfico de Frutas")
    plt.xlabel("Frutas")
    plt.ylabel("Valores Digitados")

    # 4. Troquei os números do eixo X pelos nomes reais das frutas
    plt.xticks(xs, frutas)

    # 5. Exibo o gráfico final na tela
    print("\nPlotando seu gráfico de barras...")
    plt.show()

else:
    print("\nNenhuma fruta foi digitada. O gráfico não foi gerado.")
```

```python
# Exercício 30 - Jogo da Forca
# Explicação:
# Usei a estrutura "set" para guardar as letras certas e erradas porque ela
# ignora automaticamente itens duplicados. Assim, se eu digitar a mesma letra duas vezes,
# o programa não contabiliza como dois erros diferentes.
# Pra mostrar os tracinhos da palavra secreta na tela, usei de novo a "List Comprehension",
# que verifica letra por letra: se a letra estiver no meu set de acertos, ele mostra a letra,
# se não estiver, ele mostra um "_".

import random

def jogar_forca():
    # Uma lista simples para sortear a palavra do jogo
    lista_palavras = ["PYTHON", "ALGORITMO", "TECNOLOGIA", "SISTEMA", "COMPUTADOR"]
    palavra_secreta = random.choice(lista_palavras)

    # Criando os conjuntos (sets) vazios
    letras_certas = set()
    letras_erradas = set()
    tentativas = 6

    print("--- Jogo da Forca ---")

    while tentativas > 0:
        # Monta a palavra escondida substituindo o que eu ainda não acertei por "_"
        palavra_oculta = [letra if letra in letras_certas else "_" for letra in palavra_secreta]

        print(f"\nPalavra: {' '.join(palavra_oculta)}")
        print(f"Tentativas restantes: {tentativas}")

        # Mostra as letras erradas só se a lista não estiver vazia
        if letras_erradas:
            print(f"Erros: {', '.join(letras_erradas)}")

        # Condição de vitória: se não tiver mais "_" na palavra montada, ganhei
        if "_" not in palavra_oculta:
            print("\nBoa! Você adivinhou a palavra:", palavra_secreta)
            break

        chute = input("Digite uma letra: ").upper()

        # Validação pra garantir que digitei só uma letra de verdade
        if len(chute) != 1 or not chute.isalpha():
            print("Entrada inválida! Digita só uma letra.")
            continue

        # Verifica se já tentei essa letra antes
        if chute in letras_certas or chute in letras_erradas:
            print("Você já tentou essa letra. Tenta outra!")
            continue

        # Testa se acertei ou errei
        if chute in palavra_secreta:
            letras_certas.add(chute)
        else:
            letras_erradas.add(chute)
            tentativas -= 1

    # Se sair do while porque as tentativas zeraram, mostra a derrota
    if tentativas == 0:
        print(f"\nDeu ruim! Você foi enforcado. A palavra era: {palavra_secreta}")

# Chama a função para o jogo rodar
jogar_forca()
```

```python
# Exercício 31 - Ordenação por Bolha (Bubble Sort)
# Explicação:
# Implementei o Bubble Sort com uma otimização pra ele ficar mais eficiente.
# Criei uma variável chamada 'trocou'. Se o código der uma volta inteira na lista
# e não precisar trocar nenhum número de lugar, significa que tudo já está ordenado.
# Quando isso acontece, o 'break' encerra o trabalho antes da hora e economiza processamento.
# Outro detalhe legal que usei foi a troca direta do Python (lista[j], lista[j+1] = lista[j+1], lista[j]).
# Isso inverte as posições numa linha só, sem precisar criar uma terceira variável temporária.

def ordenacao_bolha(lista):
    tamanho = len(lista)

    # O loop de fora controla quantas vezes vamos passar pela lista
    for i in range(tamanho):
        trocou = False

        # O loop de dentro vai empurrando o maior número pro final.
        # O "- i - 1" serve pra não testar os últimos números que já foram ordenados nas voltas anteriores.
        for j in range(0, tamanho - i - 1):

            # Se o número atual for maior que o próximo, eles trocam de lugar
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]
                trocou = True

        # Se terminei a volta e ninguém trocou de lugar, a lista já tá pronta!
        if not trocou:
            break

    return lista

# --- Testando a implementação ---
numeros_baguncados = [64, 34, 25, 12, 22, 11, 90]

print("Lista original:", numeros_baguncados)
numeros_arrumados = ordenacao_bolha(numeros_baguncados)
print("Lista ordenada:", numeros_arrumados)
```

```python
# Exercício 32 - Jogo da Velha
# Explicação:
# Pra organizar o Jogo da Velha e deixar o código limpo, dividi o trabalho em funções.
# O tabuleiro não passa de uma lista com 9 espaços, que começam vazios (" ").
# O truque que eu mais gostei de fazer foi a função de verificação: eu já deixei mapeado
# em uma lista todos os jeitos de ganhar (linhas, colunas e diagonais) usando os índices.
# No loop principal, o código só vai revezando a vez do jogador ("X" ou "O"), tratando os
# erros caso a pessoa digite uma letra ou escolha um espaço que já está ocupado.

def imprimir_tabuleiro(tab):
    # Imprime os itens da lista no formato do jogo
    print(f"\n {tab[0]} | {tab[1]} | {tab[2]} ")
    print("---+---+---")
    print(f" {tab[3]} | {tab[4]} | {tab[5]} ")
    print("---+---+---")
    print(f" {tab[6]} | {tab[7]} | {tab[8]} \n")

def verificar_vencedor(tab, jogador):
    # Todas as combinações possíveis de índices para fechar o jogo
    combinacoes_vitoria = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8], # Linhas
        [0, 3, 6], [1, 4, 7], [2, 5, 8], # Colunas
        [0, 4, 8], [2, 4, 6]             # Diagonais
    ]
    # Passa por cada combinação e checa se as 3 posições têm o símbolo do mesmo jogador
    for c in combinacoes_vitoria:
        if tab[c[0]] == tab[c[1]] == tab[c[2]] == jogador:
            return True
    return False

def jogar_velha():
    # Cria os 9 espaços vazios na lista
    tabuleiro = [" "] * 9
    jogador_atual = "X"
    jogadas = 0

    print("--- Jogo da Velha ---")
    print("Use os números de 1 a 9 para escolher a posição no tabuleiro:")
    # Um mini tabuleiro numerado só pra pessoa entender onde cada número cai
    imprimir_tabuleiro(["1", "2", "3", "4", "5", "6", "7", "8", "9"])

    # O jogo dura no máximo 9 rodadas
    while jogadas < 9:
        imprimir_tabuleiro(tabuleiro)
        print(f"Vez do jogador {jogador_atual}!")

        try:
            # O usuário digita de 1 a 9, mas na lista os índices são de 0 a 8, por isso o "-1"
            escolha = int(input("Escolha uma posição livre (1-9): ")) - 1

            # Checa se o número está dentro do tabuleiro válido
            if escolha < 0 or escolha > 8:
                print("Opa, posição inválida! Escolha um número de 1 a 9.")
                continue

            # Checa se alguém já não jogou naquela posição
            if tabuleiro[escolha] != " ":
                print("Esse espaço já tá ocupado. Tenta outro!")
                continue

        except ValueError:
            print("Entrada inválida! Digita um número.")
            continue

        # Marca a jogada no tabuleiro e contabiliza a rodada
        tabuleiro[escolha] = jogador_atual
        jogadas += 1

        # Testa se o jogador da vez ganhou com essa última jogada
        if verificar_vencedor(tabuleiro, jogador_atual):
            imprimir_tabuleiro(tabuleiro)
            print(f"Acabou! O jogador {jogador_atual} amassou e venceu o jogo!")
            return # Encerra a função (e o jogo) na hora

        # Troca a vez: se era X vira O. Se era O, vira X.
        jogador_atual = "O" if jogador_atual == "X" else "X"

    # Se saiu do laço das 9 jogadas e não caiu no return da vitória lá em cima, empatou
    imprimir_tabuleiro(tabuleiro)
    print("Deu velha! O jogo empatou.")

# Roda o jogo
jogar_velha()
```

```python
# Exercício 33 - Verificação de Anagramas

# Explicação:
# Nesse exercício eu utilizei o input() para receber duas palavras.
# Depois, utilizei a função sorted() para organizar as letras das duas
# palavras em ordem alfabética. Após isso, comparei as duas listas de letras.
# Se elas forem iguais, significa que as palavras possuem as mesmas letras
# em quantidades iguais, apenas em uma ordem diferente, sendo consideradas
# anagramas. Caso contrário, as palavras não são anagramas.

palavra1 = input("Digite a primeira palavra: ")  # Pede a primeira palavra

palavra2 = input("Digite a segunda palavra: ")  # Pede a segunda palavra

letras1 = sorted(palavra1)  # Organiza as letras da primeira palavra

letras2 = sorted(palavra2)  # Organiza as letras da segunda palavra

if letras1 == letras2:  # Compara as letras das duas palavras
    print("As palavras são anagramas.")  # Mostra que as palavras são anagramas

else:  # Caso as letras sejam diferentes
    print("As palavras não são anagramas.")  # Mostra que não são anagramas
```

```python
# Exercício 34 - Contador de Vogais

# Explicação:
# Nesse exercício eu utilizei o input() para receber uma frase ou palavra
# digitada pelo usuário. Depois, percorri cada caractere do texto utilizando
# um laço de repetição for. A cada letra, o programa verifica se ela é uma
# vogal (a, e, i, o ou u). Quando encontra uma vogal, o contador é aumentado.
# Ao final, o programa exibe a quantidade total de vogais encontradas no texto.

texto = input("Digite uma palavra ou frase: ")  # Pede uma palavra ou frase ao usuário

contador = 0  # Começa a contagem de vogais com 0

for letra in texto:  # Percorre cada letra do texto

    if letra == "a" or letra == "e" or letra == "i" or letra == "o" or letra == "u":  # Verifica se a letra é uma vogal
        contador += 1  # Aumenta a contagem de vogais

print("Quantidade de vogais:", contador)  # Mostra a quantidade de vogais encontradas
```

```python
# Exercício 35 - Números Fibonacci Até N
# Explicação:
# Nesse exercício eu utilizei o input() para receber um número limite.
# Depois, criei duas variáveis para guardar os primeiros valores da sequência
# de Fibonacci. A cada repetição, o programa soma os dois valores anteriores
# para gerar o próximo número. Além de exibir a sequência, o programa também
# mostra as contas realizadas para chegar em cada resultado.

limite = int(input("Digite um número limite: "))  # Pede o limite da sequência

anterior = 0  # Define o primeiro número da sequência
atual = 1  # Define o segundo número da sequência

print("Sequência de Fibonacci:")  # Mostra o título da sequência

print(anterior)  # Mostra o primeiro número
print(atual)  # Mostra o segundo número

while anterior + atual <= limite:  # Continua enquanto o próximo número não ultrapassar o limite

    proximo = anterior + atual  # Soma os dois números anteriores para encontrar o próximo

    print(anterior, "+", atual, "=", proximo)  # Mostra a soma realizada

    anterior = atual  # Passa o número atual para a posição anterior

    atual = proximo  # Passa o próximo número para a posição atual
```

```python
# Exercício 36 - Gerador de Senhas

# Explicação:
# Nesse exercício eu criei um gerador de senhas aleatórias utilizando letras
# e números. Primeiro, criei uma lista contendo todos os caracteres que podem
# ser usados na senha.
#
# Depois, utilizei o input() para receber o tamanho da senha que o usuário
# deseja criar. Em seguida, utilizei um laço de repetição for para escolher
# aleatoriamente cada caractere que fará parte da senha.
#
# A cada repetição, um novo caractere é escolhido e adicionado à variável
# que armazena a senha. Ao final, o programa exibe a senha gerada.

import random  # Importa a biblioteca para escolher caracteres aleatórios

caracteres = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"  # Define os caracteres que podem aparecer na senha

tamanho = int(input("Digite o tamanho da senha: "))  # Pede o tamanho da senha

senha = ""  # Cria uma variável vazia para guardar a senha

for i in range(tamanho):  # Repete de acordo com o tamanho escolhido

    caractere = random.choice(caracteres)  # Escolhe um caractere aleatório

    senha += caractere  # Adiciona o caractere escolhido à senha

print("Senha gerada:", senha)  # Mostra a senha gerada
```

```python
# Exercício 37 - Funções Recursivas

# Explicação:
# Nesse exercício eu criei uma função recursiva para calcular o fatorial
# de um número. Uma função recursiva é uma função que chama ela mesma
# durante sua execução.
#
# Primeiro, o programa recebe um número informado pelo usuário.
# Depois, a função verifica se o número é igual a 0 ou 1, pois nesses casos
# o resultado do fatorial é 1, encerrando a repetição.
#
# Caso o número seja maior que 1, a função multiplica o número pelo fatorial
# do número anterior, chamando ela mesma novamente até chegar ao caso final.
# Ao terminar os cálculos, o resultado do fatorial é exibido na tela.

def fatorial(numero):  # Cria uma função para calcular o fatorial

    if numero == 0 or numero == 1:  # Verifica se o número é 0 ou 1
        return 1  # Retorna 1, pois o fatorial de 0 e 1 é 1

    else:  # Caso o número seja maior que 1
        return numero * fatorial(numero - 1)  # Multiplica o número pelo fatorial do número anterior

numero = int(input("Digite um número: "))  # Pede um número ao usuário

resultado = fatorial(numero)  # Chama a função e guarda o resultado

print("Fatorial de", numero, "é:", resultado)  # Mostra o resultado do fatorial
```

```python
# Exercício 40 - Lista de Compras

# Explicação:
# Nesse exercício eu criei um programa para gerenciar uma lista de compras.
# Primeiro, criei uma lista vazia onde os produtos adicionados pelo usuário
# serão armazenados.
#
# Depois, utilizei um laço de repetição para manter o programa funcionando,
# permitindo que o usuário escolha entre adicionar um item, remover um item
# ou visualizar a lista de compras.
#
# Quando o usuário escolhe adicionar, o produto é colocado na lista.
# Quando escolhe remover, o programa procura o item e remove da lista.
# Na opção listar, todos os produtos cadastrados são exibidos na tela.

compras = []  # Cria uma lista vazia para guardar os itens

opcao = 0  # Define a opção inicial

while opcao != 4:  # Continua o programa enquanto a opção não for 4

    print("\n1 - Adicionar item")  # Mostra a opção de adicionar
    print("2 - Remover item")  # Mostra a opção de remover
    print("3 - Listar compras")  # Mostra a opção de listar
    print("4 - Sair")  # Mostra a opção de sair

    opcao = int(input("Escolha uma opção: "))  # Pede ao usuário para escolher uma opção

    if opcao == 1:  # Verifica se o usuário escolheu adicionar

        item = input("Digite o item que deseja adicionar: ")  # Pede o item que será adicionado

        compras.append(item)  # Adiciona o item à lista

        print("Item adicionado!")  # Mostra uma mensagem de confirmação

    elif opcao == 2:  # Verifica se o usuário escolheu remover

        item = input("Digite o item que deseja remover: ")  # Pede o item que será removido

        if item in compras:  # Verifica se o item está na lista

            compras.remove(item)  # Remove o item da lista

            print("Item removido!")  # Mostra uma mensagem de confirmação

        else:  # Caso o item não esteja na lista

            print("Item não encontrado!")  # Mostra uma mensagem informando que o item não existe

    elif opcao == 3:  # Verifica se o usuário escolheu listar as compras

        print("Lista de compras:")  # Mostra o título da lista

        for item in compras:  # Percorre cada item da lista

            print(item)  # Mostra cada item da lista

    elif opcao == 4:  # Verifica se o usuário escolheu sair

        print("Programa encerrado!")  # Mostra a mensagem de encerramento

    else:  # Caso seja escolhida uma opção diferente das disponíveis

        print("Opção inválida!")  # Mostra uma mensagem de erro
```

```python
# Exercício 41 - Remover Duplicatas

# Explicação:
# Nesse exercício eu criei um programa que remove os números repetidos
# de uma lista. Primeiro, criei uma lista contendo alguns números,
# inclusive alguns repetidos. Depois, criei uma nova lista vazia para
# armazenar apenas os números que ainda não apareceram.
#
# Em seguida, utilizei um laço de repetição para percorrer todos os
# elementos da lista original. A cada repetição, o programa verifica
# se o número já está na nova lista. Se não estiver, ele é adicionado
# utilizando a concatenação de listas. Ao final, a nova lista contém
# apenas os números sem repetição.

# Lista original
numeros = [5, 2, 8, 2, 4, 5, 9, 8, 1]

# Lista sem duplicatas
sem_duplicatas = []

# Percorre a lista
for numero in numeros:

    # Verifica se o número ainda não existe na nova lista
    if numero not in sem_duplicatas:

        # Adiciona o número utilizando concatenação
        sem_duplicatas = sem_duplicatas + [numero]

print("Lista original:")
print(numeros)

print("Lista sem duplicatas:")
print(sem_duplicatas)
```

```python
# Exercício 42 - Números Primos até N

# Explicação:
# Nesse exercício eu utilizei o input() para receber um número informado
# pelo usuário. Depois, utilizei um laço de repetição para verificar todos
# os números de 2 até o número digitado. Para cada número, o programa verifica
# se ele possui algum divisor além de 1 e dele mesmo. Se não possuir,
# ele é considerado um número primo e é exibido na tela.

# Recebe do usuário o número limite
limite = int(input("Digite um número: "))

# Exibe uma mensagem informando o que será mostrado
print("Números primos até", limite, ":")

# Percorre todos os números de 2 até o número informado
for numero in range(2, limite + 1):

    # No início, considera que o número é primo
    primo = True

    # Testa todos os possíveis divisores
    for divisor in range(2, numero):

        # Verifica se a divisão é exata
        if numero % divisor == 0:

            # Se encontrar um divisor, o número não é primo
            primo = False

    # Se continuar sendo primo, exibe o número
    if primo == True:

        print(numero)
```

```python
# Exercício 44 - Fibonacci em Nível

# Explicação:
# Nesse exercício eu utilizei o input() para receber a quantidade de
# termos da sequência de Fibonacci que o usuário deseja visualizar.
# Depois, criei duas variáveis para armazenar os dois primeiros números
# da sequência. Em seguida, utilizei um laço de repetição para gerar
# a quantidade de termos informada pelo usuário. A cada repetição,
# o programa exibe o número atual, calcula o próximo número da sequência
# e atualiza os valores para continuar o processo.

# Recebe a quantidade de termos da sequência
quantidade = int(input("Digite a quantidade de termos: "))

# Primeiro número da sequência
anterior = 0

# Segundo número da sequência
atual = 1

# Exibe uma mensagem
print("Sequência de Fibonacci:")

# Repete a quantidade de vezes informada
for i in range(quantidade):

    # Exibe o número atual da sequência
    print(anterior)

    # Calcula o próximo número
    proximo = anterior + atual

    # Atualiza o valor do número anterior
    anterior = atual

    # Atualiza o valor do número atual
    atual = proximo
```

```python
# Exercício 45 - Par ou Ímpar em Lista

# Explicação:
# Nesse exercício eu utilizei uma lista de números e percorri todos os
# seus elementos utilizando um laço de repetição. Para cada número,
# utilizei o operador % (resto da divisão) para verificar se ele é
# par ou ímpar.
#
# Se o resto da divisão por 2 for igual a 0, o número é par.
# Caso contrário, o número é ímpar. Ao final, o programa informa a
# classificação de cada número da lista.

# Lista de números
numeros = [8, 3, 10, 7, 12, 5, 18, 9]

# Percorre todos os números da lista
for numero in numeros:

    # Verifica se o número é par
    if numero % 2 == 0:

        # Exibe que o número é par
        print(numero, "é par")

    # Caso contrário, o número é ímpar
    else:

        # Exibe que o número é ímpar
        print(numero, "é ímpar")
```

```python
# Exercício 46 - Calcular Média e Desvio Padrão

# Explicação:
# Nesse exercício eu utilizei uma lista de números para calcular a média
# e o desvio padrão. Primeiro, somei todos os números da lista e dividi
# pela quantidade de elementos para encontrar a média.
#
# Depois, calculei a diferença entre cada número e a média. Essa diferença
# foi elevada ao quadrado e somada. Em seguida, dividi essa soma pela
# quantidade de números para encontrar a variância.
#
# Por último, calculei a raiz quadrada da variância utilizando o operador
# de potência (** 0.5), encontrando o desvio padrão.

# Lista de números
numeros = [5, 8, 10, 6, 9]

# Variável para armazenar a soma dos números
soma = 0

# Variável para contar a quantidade de números
quantidade = 0

# Percorre toda a lista
for numero in numeros:

    # Soma os números
    soma = soma + numero

    # Conta quantos números existem
    quantidade = quantidade + 1

# Calcula a média
media = soma / quantidade

# Exibe a média
print("Média:", media)

# Variável para armazenar a soma das diferenças ao quadrado
soma_diferencas = 0

# Percorre novamente todos os números
for numero in numeros:

    # Calcula a diferença entre o número e a média
    diferenca = numero - media

    # Eleva a diferença ao quadrado
    quadrado = diferenca ** 2

    # Soma os quadrados
    soma_diferencas = soma_diferencas + quadrado

# Calcula a variância
variancia = soma_diferencas / quantidade

# Calcula o desvio padrão (raiz quadrada da variância)
desvio = variancia ** 0.5

# Exibe o desvio padrão
print("Desvio padrão:", desvio)
```

```python
# Exercício 47 - Contar Números em Lista

# Explicação:
# Nesse exercício eu utilizei uma lista de números e contei quantas vezes
# cada número aparece. Primeiro, percorri toda a lista. Para evitar contar
# o mesmo número mais de uma vez, verifiquei se ele já havia sido contado.
# Se ainda não tivesse sido contado, percorri novamente a lista para contar
# quantas vezes ele aparecia. No final, exibi o número e a quantidade de
# vezes que ele foi encontrado.

# Lista de números
numeros = [2, 5, 2, 8, 5, 2, 9, 8, 1]

# Lista para guardar os números que já foram contados
contados = []

# Percorre todos os números da lista
for numero in numeros:

    # Verifica se o número ainda não foi contado
    if numero not in contados:

        # Variável para contar quantas vezes o número aparece
        contador = 0

        # Percorre novamente toda a lista
        for valor in numeros:

            # Verifica se o número é igual ao valor da lista
            if numero == valor:

                # Aumenta o contador
                contador = contador + 1

        # Exibe o resultado
        print(numero, "aparece", contador, "vez(es).")

        # Guarda o número para não contá-lo novamente
        contados = contados + [numero]
```

```python
# Exercício 48 - Ordenação por Inserção (Insertion Sort)

# Explicação:
# Nesse exercício eu utilizei o algoritmo Insertion Sort para ordenar
# uma lista de números em ordem crescente.
#
# O algoritmo funciona como se estivéssemos organizando cartas na mão.
# Ele começa considerando que o primeiro número já está na posição correta.
# Depois, pega o próximo número e o compara com os anteriores.
# Se o número for menor, os outros números são deslocados para a direita
# até que seja encontrada a posição correta para inseri-lo.
#
# Esse processo é repetido até que toda a lista esteja ordenada.

# Lista de números
numeros = [8, 3, 10, 1, 6]

# Exibe a lista antes da ordenação
print("Lista inicial:")
print(numeros)

# Percorre a lista a partir do segundo elemento
for i in range(1, len(numeros)):

    # Guarda o número que será inserido
    atual = numeros[i]

    # Guarda a posição anterior
    j = i - 1

    # Enquanto existir posição anterior
    # e o número anterior for maior que o atual
    while j >= 0 and numeros[j] > atual:

        # Desloca o número maior uma posição para a direita
        numeros[j + 1] = numeros[j]

        # Vai para a posição anterior
        j = j - 1

    # Coloca o número na posição correta
    numeros[j + 1] = atual

    # Exibe a lista após cada inserção
    print(numeros)

# Exibe a lista ordenada
print("Lista ordenada:")
print(numeros)
```
